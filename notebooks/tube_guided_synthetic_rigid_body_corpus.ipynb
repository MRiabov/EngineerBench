# Tube-Guided Synthetic Rigid-Body Corpus Generator

This notebook is the interactive front-end. The heavy lifting lives in `dataset/synthetic/tube_guided_synthetic_rigid_body_corpus/` so the notebook stays readable.

Edit the benchmark bundle path and route points, then run the single code cell below. The generator will:

- build the scratch tube scaffold,
- run the batched MuJoCo / Genesis acceptance loop,
- capture and prune the wall-contact cloud,
- write planner and coder row bundles with `.solution/` overlays,
- optionally promote the result into `dataset/data/seed/artifacts/**`.


```python
import sys
from pathlib import Path

ROOT = Path.cwd()
for candidate in (ROOT, *ROOT.parents):
    if (candidate / 'specs' / 'desired_architecture.md').exists():
        ROOT = candidate
        break

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from dataset.synthetic.tube_guided_synthetic_rigid_body_corpus import (  # noqa: E402
    ScenarioConfig,
    default_route_points,
    main,
)

config = ScenarioConfig(
    scenario_id='tube-guided-synthetic-001',
    source_seed_bundle_dir=ROOT / 'dataset' / 'data' / 'seed' / 'artifacts' / 'engineer_coder' / 'ec-002-low-friction-cube',
    planner_row_id='ep-synth-001',
    coder_row_id='ec-synth-001',
    route_points=default_route_points(),
    scratch_root=ROOT / 'tmp' / 'tube_guided_synthetic_corpus',
    backfill_source_solution=False,
    promote_to_dataset=False,
)

try:
    summary = main(config)
except ModuleNotFoundError as exc:
    summary = {
        'error': type(exc).__name__,
        'message': str(exc),
        'hint': 'Install the optional simulation/CAD dependencies before running the full export.',
    }
summary
```

If the run looks good, switch `promote_to_dataset=True` in the code cell and rerun to copy the staged bundles into the canonical seed dataset paths and update the role-based row JSON files.
